```python
#Bisection method
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x)**3 + x - 1

clist = []
a=0.1
b=1
tol = 10**(-8)

if f(a)*f(b) >0:
    print('choose different points')

else:
    while np.abs(b-a)/2>tol:
        c = 0.5*(a+b)
        clist.append(c)
        if f(a)*f(c) < 0:
            b = c

        else:
            a = c



    print('root = ',c)
    print('f(c) = ',f(c))

x = np.arange(0,1,0.1)
clist = np.array(clist)

plt.plot(x,f(x), label='f(x) = x**3 + x - 1')
plt.scatter(clist,f(clist),color='green', label='c values')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.show()


plt.figure()
plt.plot(clist)
plt.xlabel('Interation number')
plt.ylabel('c')
plt.show()
```

```python
#Newton Raphson method
import numpy as np
import matplotlib.pyplot as plt

x0 = 0.1
tol = 10**(-8)
val = 1
clist1=[]
def fd(x):
  return 3*x**2 + 1

if np.abs(f(x0)) < tol:
  print('root =',x0)

else:
  while np.abs(val)>tol:
    x = x0 - f(x0)/fd(x0)
    val = (x - x0)
    x0 = x
    clist1.append(x0)
  print('root = ',x)

plt.figure()
plt.plot(clist1)
plt.xlabel('Interation number')
plt.ylabel('c')
plt.show()
```

```python
#Secant method
import numpy as np
import matplotlib.pyplot as plt

x0 = 0.1
x1 = 0.2
tol = 10**(-8)
val = 1
clist2 = []

if np.abs(f(x0)) < tol:
  print('root =',x0)

elif np.abs(f(x1)) < tol:
  print('root =',x1)

else:
  while np.abs(val)>tol:
    fd = (f(x1) - f(x0))/(x1 - x0)
    x = x1 - f(x1)/fd
    val = (x - x1)
    clist2.append(x)
    x0=x1
    x1 =x


  print('root = ',x1)

plt.figure()
plt.plot(clist2)
plt.xlabel('Interation number')
plt.ylabel('c')
plt.show()
```

```python
#Inverse Quadratic Interpolation
import numpy as np
import matplotlib.pyplot as plt

x0 = 0.1
x1 = 0.15
x2 = 0.25
tol = 10**(-8)
val = 1
clist3 = []

if np.abs(f(x0)) < tol:
  print('root =',x0)

elif np.abs(f(x1)) < tol:
  print('root =',x1)

elif np.abs(f(x2)) < tol:
  print('root =',x2)

else:
  while np.abs(val)>tol:
    q = f(x0)/f(x1)
    r = f(x2)/f(x1)
    s = f(x2)/f(x0)
    x3 = x2 - ((r*(r - q)*(x2 - x1)) + s*(1-r)*(x2 - x0))/((q-1)*(r-1)*(s-1))
    clist3.append(x3)
    val = (x3 - x2)
    x0 = x1
    x1 = x2
    x2 = x3


  print('root = ',x2)

plt.figure()
plt.plot(clist3)
plt.xlabel('Interation number')
plt.ylabel('c')
plt.show()
```
