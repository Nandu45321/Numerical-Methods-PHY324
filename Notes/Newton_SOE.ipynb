```python
import numpy as np


def func(x):
    return -1 * np.array([
        x[1] - x[0]**3,
        x[0]**2 + x[1]**2 - 1
    ])


# Initial guess
x = np.array([1.0, 2.0])

# Jacobian matrix
a = np.array([[-3 * x[0]**2, 1],[2 * x[0], 2 * x[1]]], dtype=float)

# Function vector
f = func(x)

# Number of variables
n = a.shape[0]

# Tolerance
tol = n * 1e-2

# Difference between successive x vectors
dx = np.ones(n)


while np.sum(dx) > tol:

    xprev = x.copy()

    # Gaussian elimination with partial pivoting
    for j in range(n - 1):

        # Find maximum element for pivoting
        I = j + np.argmax(np.abs(a[j:n, j]))

        # Row swapping
        if I != j:
            a[[j, I], :] = a[[I, j], :]
            f[[j, I]] = f[[I, j]]

        # Elimination
        for i in range(j + 1, n):

            # Multiplication factor
            mult = a[i, j] / a[j, j]

            for k in range(j, n):
                a[i, k] = a[i, k] - mult * a[j, k]

            f[i] = f[i] - mult * f[j]

    # Back substitution
    s = np.zeros(n)

    # Last variable
    s[n - 1] = f[n - 1] / a[n - 1, n - 1]

    x[n - 1] = x[n - 1] + s[n - 1]

    # Remaining variables
    for i in range(n - 2, -1, -1):

        SUM = 0.0

        for j in range(i + 1, n):
            SUM = SUM + a[i, j] * s[j]

        s[i] = (f[i] - SUM) / a[i, i]

        x[i] = x[i] + s[i]

    # Update Jacobian
    a = np.array([
        [-3 * x[0]**2, 1],
        [2 * x[0], 2 * x[1]]
    ], dtype=float)

    # Update function vector
    f = func(x)

    # Calculate difference between iterations
    dx = np.abs(xprev - x)


# Display solution
print("Solution:")
print(x)
```
