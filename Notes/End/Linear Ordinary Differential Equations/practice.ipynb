```python
import numpy as np
import matplotlib.pyplot as plt
lam = 0.5
N0 = 100
dt = 0.5
t = np.arange(0, 10 + dt, dt)
n = len(t)

N_exact = N0 * np.exp(-lam*t)
N_euler = np.zeros(n); N_euler[0] = N0
N_imp = np.zeros(n); N_imp[0] = N0
N_implicit = np.zeros(n); N_implicit[0] = N0

for i in range(n-1):
    N_euler[i+1] = N_euler[i] * (1 - lam*dt)
    N_imp[i+1] = N_imp[i] * (1 - lam*dt*(1 - lam*dt/2))
    N_implicit[i+1] = N_implicit[i] / (1 + lam * dt)

plt.figure(figsize=(7, 4))
plt.plot(t, N_exact, 'k--')
plt.plot(t, N_euler, 'g--')
plt.plot(t, N_imp, 'r--')
plt.plot(t, N_implicit, 'b--')
plt.show
```
