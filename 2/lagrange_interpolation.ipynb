```python
import numpy as np
import matplotlib.pyplot as plt

#define var pts
t = np.arange(0, np.pi/2, 0.01)

#interpolation nodes
x = np.array([0, np.pi/6, 2*np.pi/6, 3*np.pi/6])

#init lagrange basis polynomial
l = np.zeros((len(t), 4))

l[:, 0] = (t - x[1]) * (t - x[2]) * (t - x[3]) / ((x[0] - x[1]) * (x[0] - x[2]) * (x[0] - x[3]))
l[:, 1] = (t - x[0]) * (t - x[2]) * (t - x[3]) / ((x[1] - x[0]) * (x[1] - x[2]) * (x[1] - x[3]))
l[:, 2] = (t - x[0]) * (t - x[1]) * (t - x[3]) / ((x[2] - x[0]) * (x[2] - x[1]) * (x[2] - x[3]))
l[:, 3] = (t - x[0]) * (t - x[1]) * (t - x[2]) / ((x[3] - x[0]) * (x[3] - x[1]) * (x[3] - x[2]))


#lagrange interpolating polynomial for sin(t)
L = np.sin(x[0])*l[:,0] + np.sin(x[1])*l[:,1] + np.sin(x[2])*l[:,2] + np.sin(x[3])*l[:,3]

#plotting
plt.plot(t, np.sin(t), label = 'sin(t)')
plt.scatter(x, np.sin(x), color = 'red', marker = 'x', label = 'Interpolation nodes')
plt.plot(t, L, label = 'Lagrange Interpolating Polynomial')
plt.show()
```
