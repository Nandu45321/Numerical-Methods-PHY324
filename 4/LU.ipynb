# LU Decomposition
### Purpose
LU decomposition factorizes a matrix $A$ into the product of a lower triangular matrix $L$ and an upper triangular matrix $U$ ($A = LU$). 

Its main purpose is to make solving systems of linear equations ($Ax = b$) much more efficient when we have to solve for multiple different $b$ vectors over time. 

Instead of doing the expensive Gaussian elimination process ($O(n^3)$ operations) every single time for a new $b$, we factorize $A$ **once**. Then, we can solve two simpler triangular systems ($Ly = b$, and $Ux = y$) very quickly ($O(n^2)$ operations).


```python
import numpy as np

def lu_decomposition(A):
    n = A.shape[0]
    L = np.eye(n)
    U = A.copy()
    
    for i in range(n):
        for j in range(i+1, n):
            factor = U[j, i] / U[i, i]
            U[j] -= factor * U[i]
            L[j, i] = factor
            
    return L, U

def solve_lu(L, U, b):
    # 1. Forward substitution (Solve L y = b)
    y = np.linalg.solve(L, b)
        
    # 2. Backward substitution (Solve U x = y)
    x = np.linalg.solve(U, y)
        
    return x

A = np.array([[3.0, 2.0, -4.0], [2.0, 3.0, 3.0], [5.0, -3.0, 1.0]])
b = np.array([3.0, 15.0, 14.0])

# Factorize once
L, U = lu_decomposition(A)

# Solve for x
x = solve_lu(L, U, b)

print("L matrix:\n", L)
print("\nU matrix:\n", U)
print("\nVerify L @ U == A:\n", np.allclose(L @ U, A))
print("\nSolution x:", x)
```
