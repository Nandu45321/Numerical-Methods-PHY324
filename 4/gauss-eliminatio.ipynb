# 3x3 Gaussian Elimination
Standard Gaussian Elimination implementation with Forward Elimination and Back Substitution.


```python
import numpy as np

def gauss_elimination(A, b):
    n = len(b)
    # Forward elimination
    for i in range(n):
        for j in range(i+1, n):
            factor = A[j, i] / A[i, i]
            A[j] -= factor * A[i]
            b[j] -= factor * b[i]
            
    # Back substitution
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        known_chunk = np.dot(A[i, i+1:], x[i+1:])
        x[i] = (b[i] - known_chunk) / A[i, i]
        
    return x

# Example 3x3 system
A = np.array([[3.0, 2.0, -4.0], [2.0, 3.0, 3.0], [5.0, -3.0, 1.0]])
b = np.array([3.0, 15.0, 14.0])

x = gauss_elimination(A.copy(), b.copy())
print("Solution x:", x)
```

look at what your equations look like after Forward Elimination is finished for a 3x3 matrix:

*   **Row 0:** A[0,0]*x0 + A[0,1]*x1 + A[0,2]*x2 = b[0]
*   **Row 1:** A[1,1]*x1 + A[1,2]*x2 = b[1]  *(x0 is gone)*
*   **Row 2:** A[2,2]*x2 = b[2] *(x0 and x1 are gone)*

Because we are doing backwards substitution, the loop (i) starts at the bottom (i=2) and goes up to the top (i=0). Here is literally what those two lines of code do at each step:

---

### Step 1: Bottom Row (i = 2)
Your math equation is: A[2,2]*x2 = b[2]. You need to solve for x2.
*   "i+1" means index 3. There is nothing at index 3. 
*   Because there are no variables to the right of x2, `known_chunk` is just 0.
*   The second line becomes: `x[2] = (b[2] - 0) / A[2,2]`
*   **Math translated:** x2 = b[2] / A[2,2]

---

### Step 2: Middle Row (i = 1)
Your math equation is: A[1,1]*x1 + A[1,2]*x2 = b[1]. You need to solve for x1.
You already know the answer for x2 from the step above.
*   `A[1, 2:]` grabs all the coefficients to the right of your current variable. That's just A[1,2].
*   `x[2:]` grabs all the variables you already solved for. That's just x2.
*   `known_chunk = np.dot(...)` multiplies them: A[1,2] * x2
*   The second line becomes: `x[1] = (b[1] - known_chunk) / A[1,1]`
*   **Math translated:** x1 = (b[1] - A[1,2]*x2) / A[1,1]

---

### Step 3: Top Row (i = 0)
Your math equation is: A[0,0]*x0 + A[0,1]*x1 + A[0,2]*x2 = b[0]. You need to solve for x0.
You already know the answers for x1 and x2.
*   `A[0, 1:]` grabs all coefficients to the right. That is a list: `[ A[0,1], A[0,2] ]`
*   `x[1:]` grabs all variables you already solved for. That is a list: `[ x1, x2 ]`
*   `known_chunk = np.dot(...)` pairs them up, multiplies them, and adds them. It calculates: (A[0,1] * x1) + (A[0,2] * x2).
*   The second line becomes: `x[0] = (b[0] - known_chunk) / A[0,0]`
*   **Math translated:** x0 = (b[0] - (A[0,1]*x1 + A[0,2]*x2)) / A[0,0]

---
**Summary:**
`[i+1:]` is just Python's way of saying "grab every number that comes after my current position". `np.dot` just multiplies those numbers by the answers we already found and adds them up.
