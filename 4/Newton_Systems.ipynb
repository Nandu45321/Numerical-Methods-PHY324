# Newton's Method for Systems of Nonlinear Equations
Normal Newton's method uses the derivative (slope) to find roots of 1 equation. 
But what if we have multiple equations interacting with multiple variables?
We use the Jacobian Matrix (which is basically a matrix of all the partial derivatives). At every step, we evaluate the Jacobian, and solve a linear system of equations (using our good old Gaussian Elimination/LU trick) to find the perfect step direction for all variables at once.

### The Math
For a system of equations $\mathbf{F}(\mathbf{x}) = \mathbf{0}$, the multi-variable Taylor expansion gives us the Newton iteration:
$$ \mathbf{x}_{i+1} = \mathbf{x}_i - [\mathbf{J}(\mathbf{x}_i)]^{-1} \mathbf{F}(\mathbf{x}_i) $$
where $\mathbf{J}(\mathbf{x}_i)$ is the Jacobian matrix evaluated at the current guess.

We explicitly compute the inverse of the Jacobian matrix, $\mathbf{J}^{-1}$, and multiply it by $\mathbf{F}(\mathbf{x}_i)$.

We update our guess using the standard inverse formula:
$$ \mathbf{x}_{i+1} = \mathbf{x}_i - [\mathbf{J}(\mathbf{x}_i)]^{-1} \mathbf{F}(\mathbf{x}_i) $$

```python
import numpy as np

# We have two equations we want to solve simultaneously:
# 1) x1 - x0^3 = 0
# 2) x0^2 + x1^2 - 1 = 0
def F(x):
    return np.array([
        x[1] - x[0]**3,
        x[0]**2 + x[1]**2 - 1
    ])

# The Jacobian is a matrix of the partial derivatives of those equations
def Jacobian(x):
    return np.array([
        [-3 * x[0]**2,  1],
        [ 2 * x[0],     2 * x[1]]
    ])

def newton_system(F, Jacobian, initial_guess, tol=1e-6, max_iter=100):
    x = np.array(initial_guess, dtype=float)
    
    for i in range(max_iter):
        J = Jacobian(x)
        fx = F(x)
        
        # Find the inverse of the Jacobian
        J_inv = np.linalg.inv(J)
        
        # Calculate the step: - J_inv * F(x)
        step = -(J_inv @ fx)
        
        # Update x: x = x + step
        x = x + step
        
        # If the step was tiny, we have converged!
        if np.linalg.norm(step) < tol:
            print(f"Converged in {i+1} iterations.")
            return x
            
    return x

# Example usage
initial_guess = [1.0, 2.0]
solution = newton_system(F, Jacobian, initial_guess)
print("Solution:", solution)
```
