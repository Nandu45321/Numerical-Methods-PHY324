```python
import matplotlib.pyplot as plt
import numpy as np
def decay(init_particles, lamb, time):
    return init_particles*np.exp(-lamb*time)
```

```python
init_particles = 1000
lamb = 0.1
step = 1
time = np.arange(1, 100, step)
noof_patricles = decay(init_particles, lamb, time)
```

```python
plt.plot(time, noof_patricles)
plt.xlabel("Time")
plt.ylabel("Number of Particles")
plt.title("Nuclear Decay")
plt.show()
```

## Simple Euler

```python
N0 = 1000
lamb = 0.1
t_end = 100
dt = 5
t = np.arange(0, t_end, dt)
N = np.zeros(len(t))
N[0] = N0

for i in range(len(t)-1):
    N[i+1] = N[i] * (1 - lamb * dt)

plt.plot(t, N)
plt.plot(time, noof_patricles)
plt.xlabel("Time")
plt.ylabel("Number of Particles")
plt.title("Nuclear Decay")
plt.show()
```

## Improved Euler Method

```python
N0 = 1000
lamb = 0.1
t_end = 100
dt = 5
t = np.arange(0, t_end, dt)
N = np.zeros(len(t))
N[0] = N0

for i in range(len(t)-1):
    N[i+1] = N[i] * (1 - lamb * dt * (1 - lamb * dt * 0.5))

plt.plot(t, N)
plt.plot(time, noof_patricles)
plt.xlabel("Time")
plt.ylabel("Number of Particles")
plt.title("Nuclear Decay")
plt.show()
```

## TK4 Method

```python
N0 = 1000
lam = 0.1
t_end = 100
dt = 1
t = np.arange(0, t_end, dt)
n = len(t)
N = np.zeros(n)
N[0] = N0

for i in range(n-1):
    k1 = -lam * N[i]
    k2 = -lam * (N[i] + dt*k1/2)
    k3 = -lam * (N[i] + dt*k2/2)
    k4 = -lam * (N[i] + dt*k3)

    N[i+1] = N[i] + dt/6 * (k1 + 2*k2 + 2*k3 + k4)

plt.plot(t, N)
plt.plot(time, noof_patricles)
plt.xlabel("Time")
plt.ylabel("Number of Particles")
plt.title("Nuclear Decay")
plt.show()
```
