#### Use simple euler: y(t + h) = y(t) + hf(y, t)

### x' = v
### mv' + yv + kx = 0

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 1
gamma = 0
k = 1
dt = 0.05
t_end = 100
t = np.arange(0, t_end, dt)
n = len(t)

v = np.zeros(n)
x = np.zeros(n)

#Need 2 initial condition to solve a 2nd order ODE
v[0] = 0
x[0] = 1
```

```python
for i in range(n-1):
    x[i+1] = x[i] + dt * v[i]
    v[i+1] = v[i] - dt * (gamma/m * v[i] + k/m * x[i])
    
```

```python

plt.style.use('dark_background')
plt.plot(t, x)
plt.xlabel("Time")
plt.ylabel("Displacement")
```

### Improved Euler

```python
m = 1
gamma = 0
k = 1
dt = 0.05
t_end = 100
t = np.arange(0, t_end, dt)
n = len(t)

v = np.zeros(n)
x = np.zeros(n)

#Need 2 initial condition to solve a 2nd order ODE
v[0] = 0
x[0] = 1
```

```python
''' Simple Euler
for i in range(n-1):
    x[i+1] = x[i] + dt/2 * (v[i] + v[i-1])
    v[i+1] = v[i] - dt * (gamma/m * v[i] + k/m * x[i])
'''    

for i in range(n-1):
    # Predictor
    x_pred = x[i] + dt * v[i]
    v_pred = v[i] - dt * (gamma/m * v[i] + k/m * x[i])
    
    # Corrector
    x[i+1] = x[i] + dt/2 * (v[i] + v_pred)
    v[i+1] = v[i] - dt/2 * ((gamma/m * v[i] + k/m * x[i]) + (gamma/m * v_pred + k/m * x_pred))
```

```python

plt.style.use('dark_background')
plt.plot(t, x)
plt.xlabel("Time")
plt.ylabel("Displacement")
```
